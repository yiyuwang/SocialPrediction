## First-Level Analysis: Subjective-Prior Model with Cue Offset

This notebooks tests the effect of cue offset by adding a `cue_offset` event regressor.

```python
from pathlib import Path
import sys

sys.path.insert(0, str(Path.cwd() / "src"))

from agentprediction import AnalysisConfig

config = AnalysisConfig()
subjects = config.load_subjects()
print("subjects in this analysis:")
print(subjects)
print(f"**** n = {len(subjects)} *****")
```

```python
from agentprediction.events import parse_subjective_prior_events
from agentprediction.first_level import fit_concatenated_first_level, fit_run_design_matrices
```

```python
model_name = "model_cue_offset"
create_design_matrix = True
include_cue_offset = True
shared_columns = [
    "obs_Pattern",
    "obs_Social",
    "fb_Pattern_Congruent",
    "fb_Social_Congruent",
    "fb_Pattern_PE",
    "fb_Social_PE",
]
n_regressors_to_save = 9
```

```python
video_key = config.load_video_key()

design_matrices = fit_run_design_matrices(
    config=config,
    model_name=model_name,
    subjects=subjects,
    event_parser=parse_subjective_prior_events,
    event_parser_kwargs={
        "video_key": video_key,
        "include_cue_offset": include_cue_offset,
    },
    create_design_matrix=create_design_matrix,
)

fit_concatenated_first_level(
    config=config,
    model_name=model_name,
    subjects=subjects,
    shared_columns=shared_columns,
    n_regressors_to_save=n_regressors_to_save,
    design_matrices=design_matrices,
)
```
